## Check missing values

```python
null = df.isnull().sum()

null[null > 0]
mispercent = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    "MisCount": df.isnull().sum(),
    "MisPercentage": mispercent
})

missing_table[missing_table["MisCount"] > 0]
```

## Cleaning the data

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df, x="Season", y="Yield_Tonnes_Ha")
plt.title("Yield Distribution Across Seasons")
plt.show()
```

The box plot helps compare the yield distribution across seasons. Kharif and Rabi have similar typical yields and greater variation, while Zaid shows a slightly lower and more compact distribution. The extreme points indicate some unusually high-yield observations.

## How does yield vary by season?

```python
season_yield = df.groupby("Season")["Yield_Tonnes_Ha"].agg(
    ["mean", "median", "min", "max"]
)

season_yield
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="Season",
    y="Yield_Tonnes_Ha",
    estimator="mean"
)

plt.title("Average Crop Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.show()
```

From the graph, Kharif has the highest average yield, followed by Rabi and Zaid. This shows that crop performance is different across seasons, and Kharif performs better in terms of average yield in this dataset.

## Which Season Has the Highest Average Profit?

```python
season_profit = df.groupby("Season")["Profit_INR"].agg(
    ["mean", "median", "min", "max"]
)

season_profit
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="Season",
    y="Profit_INR",
    estimator="mean"
)

plt.title("Average Profit by Season")
plt.xlabel("Season")
plt.ylabel("Average Profit (INR)")
plt.show()
```

## rainfall across season

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="Season",
    y="Rainfall_mm",
    estimator="mean"
)

plt.title("Average Rainfall by Season")
plt.xlabel("Season")
plt.ylabel("Rainfall (mm)")
plt.show()
```

The graph shows that Kharif has the highest average rainfall, followed by Rabi and Zaid. This clearly shows that rainfall conditions vary across the three seasons. ## Temperature across season

```python
plt.figure(figsize=(6,5))

sns.barplot(
    data=df,
    x="Season",
    y="Avg_Temperature_C",
    estimator="mean"
)

plt.title("Average Temperature by Season")
plt.xlabel("Season")
plt.ylabel("Temperature (°C)")
plt.show()
```

## Does disease/pest risk change by season?

```python
df.groupby("Season")["Disease_Pest_Risk_pct"].mean()


plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="Season",
    y="Disease_Pest_Risk_pct",
    estimator="mean"
)

plt.title("Disease and Pest Risk Across Seasons")
plt.xlabel("Season")
plt.ylabel("Risk (%)")
plt.show()
```

The graph shows that the average disease and pest risk varies across the three seasons. This indicates that agricultural risks are not the same throughout the year, and some seasons may require more attention and monitoring to manage possible disease and pest-related issues.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Rainfall_mm",
    y="Yield_Tonnes_Ha",
    hue="Season"
)

plt.title("Relationship Between Rainfall and Crop Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.show()



df["Rainfall_mm"].corr(df["Yield_Tonnes_Ha"])
```

From the scatter plot, the data points are spread across the graph without forming a clear pattern. I cannot see yield consistently increasing or decreasing as rainfall changes. This indicates that rainfall and yield do not have a strong direct relationship in this dataset. The correlation value of around 0.03 also supports this observation, showing that rainfall alone is not strongly related to crop yield.


# Does temperature relate to yield?

```python

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Avg_Temperature_C",
    y="Yield_Tonnes_Ha",
    hue="Season"
)

plt.title("Temperature vs Crop Yield")
plt.xlabel("Average Temperature (°C)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.show()
```

From the scatter plot, the points are spread across different temperature and yield values without showing a clear upward or downward pattern. This means that crop yield does not consistently increase or decrease as temperature changes. Based on the graph, there does not appear to be a strong visible relationship between temperature and crop yield in this dataset. This suggests that crop yield may depend on several factors rather than temperature alone.

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="Crop",
    y="Yield_Tonnes_Ha",
    estimator="mean"
)

plt.title("Average Yield of Different Crops")
plt.xlabel("Crop")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=45)

plt.show()
```

From the bar graph, I can see that the average yield is not the same for all crops. Some crops have higher average yields, while others have lower yields. By comparing the height of the bars, it becomes easy to identify which crops perform better in terms of average yield. This shows that crop type is also an important factor to consider when analyzing agricultural performance.
